```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt


df = pd.read_csv("diabetes.csv")
df.head()
```

```python


#Finding 0s and null
df[df['Glucose'] == 0]
df[df['BloodPressure'] == 0]
df[df['SkinThickness'] == 0]
df[df['Insulin'] == 0]
df[df['BMI'] == 0]
df[df['DiabetesPedigreeFunction'] == 0]
df[df['Age'] == 0]

df.isnull()
```

```python
from sklearn.model_selection import train_test_split

parameters = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
x = df[parameters].values
y = df['Outcome'].values

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state= 42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, f1_score, precision_score, recall_score, roc_auc_score

model = LogisticRegression(random_state= 42).fit(x_train, y_train)

y_predict = model.predict(x_test)

cm  = confusion_matrix(y_test, y_predict)

f1s = f1_score(y_test, y_predict)

prs = precision_score(y_test, y_predict)

res = recall_score(y_test, y_predict)

y_proba_std = model.predict_proba(x_test)[:, 1]
rcs = roc_auc_score(y_test, y_proba_std)

print("Confusion Matrix : \n", cm)
print("F1 Score :\t\t", f1s)
print("Precision Score :\t", prs)
print("Recall Score :\t\t", res)
print("ROC AUC Score :\t\t",rcs)
```

```python
#MIN MAX Scaling
df_minmax = pd.DataFrame()
parameters = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI',
               'DiabetesPedigreeFunction', 'Age']


df_minmax['Pregnancies']   = df['Pregnancies']/17
df_minmax['Glucose']       = df['Glucose']/199
df_minmax['BloodPressure'] = df['BloodPressure']/122
df_minmax['SkinThickness'] = df['SkinThickness']/99
df_minmax['Insulin']       = df['Insulin']/846
df_minmax['BMI']           = df['BMI']/67.1
df_minmax['DiabetesPedigreeFunction'] = df['DiabetesPedigreeFunction']/2.42
df_minmax['Age'] = df['Age']/81
df_minmax['Outcome'] = df['Outcome']

x_minmax = df_minmax[parameters].values
y_minmax = df_minmax['Outcome'].values

x_train_mm, x_test_mm, y_train_mm, y_test_mm = train_test_split(
                                               x_minmax, y_minmax, test_size=0.2, random_state=42)


model = LogisticRegression(random_state= 42).fit(x_train_mm, y_train_mm)

y_predict_mm = model.predict(x_test_mm)

cm  = confusion_matrix(y_test_mm, y_predict_mm)

f1s = f1_score(y_test_mm, y_predict_mm)

prs = precision_score(y_test_mm, y_predict_mm)

res = recall_score(y_test_mm, y_predict_mm)

y_proba_std = model.predict_proba(x_test_mm)[:, 1]
rcs = roc_auc_score(y_test, y_proba_std)


print("Confusion Matrix : \n", cm)
print("F1 Score :\t\t", f1s)
print("Precision Score :\t", prs)
print("Recall Score :\t\t", res)
print("ROC AUC Score :\t\t",rcs)

```
